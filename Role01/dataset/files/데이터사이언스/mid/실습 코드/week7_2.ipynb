```python
import pandas as pd

hw_data = {
    '지점명' : ['서울지점', '부산지점', '제주지점'],
    '컴퓨터' : [869, 578, 321],
    '모니터' : [374, 286, 148],
    '노트북' : [569, 427, 335],
    '스마트폰': [197, 89, 66]
}

# 1. DataFrame 생성 후 출력
df = pd.DataFrame(hw_data)
print(df)
print()

# 2. loc으로 서울지점 데이터 선택
print(">> loc으로 서울지점 데이터 출력")
print(df.loc[0])
print()

# 3. iloc으로 제주지점 데이터 선택
print(">> iloc으로 제주지점 데이터 출력")
print(df.iloc[2])
print()

# 4. for문으로 각 지점 총 판매량 계산
print(">> 지점별 총 판매량")
items = ['컴퓨터', '모니터', '노트북', '스마트폰']

for i in range(len(hw_data['지점명'])):
    total = 0
    for item in items:
        total += hw_data[item][i]
    print(hw_data['지점명'][i], "총 판매량:", format(total, ',') + "대")


best_branch = ''
best_total  = 0

for i in range(len(hw_data['지점명'])):
    total = 0
    for item in items:
        total += hw_data[item][i]
    if total > best_total:
        best_total  = total
        best_branch = hw_data['지점명'][i]

print(best_branch, ":", format(best_total, ',') + "대")
```

```python
import pandas as pd

df = pd.read_csv('/content/week_7_convenience_store.csv')
df['매출액'] = df['단가'] * df['판매수량']

# 행 수, 열 수
print("전체 행 수: {}행, 열 수: {}열".format(df.shape[0], df.shape[1]))
```

```python
# 지점 목록
print("지점 목록:", df['지점'].unique())

# 카테고리 목록
print("카테고리 목록:", df['카테고리'].unique())
print()

# 총 매출액
total = 0
for i in range(len(df)):
    total += df.iloc[i]['매출액']
print("전체 총 매출액:", format(total, ',') + "원")
```

```python
# 카테고리가 식품인 행
print("[ 식품 카테고리 데이터 ]")
food_df = df.loc[df['카테고리'] == '식품']
print(food_df)
```

```python
# 홍대점 음료
print("[ 홍대점 음료 데이터 ]")
hongdae_drink = df.loc[(df['지점'] == '홍대점') & (df['카테고리'] == '음료')]
print(hongdae_drink)
print(len(hongdae_drink))          # 6
print(hongdae_drink.shape[0])      # 6
```

```python
# 지점별 총 매출액
branches = list(df['지점'].unique())
branch_sales = {}

for branch in branches:
    total = 0
    for i in range(len(df)):
        if df.iloc[i]['지점'] == branch:
            total += df.iloc[i]['매출액']
    branch_sales[branch] = total

print("[ 지점별 총 매출액 ]")
for branch, sales in branch_sales.items():
    print(branch, "총 매출액:", format(sales, ',') + "원")
```

```python
# 품목별 총 판매수량
items = list(df['품목'].unique())
item_qty = {}

for item in items:
    total = 0
    for i in range(len(df)):
        if df.iloc[i]['품목'] == item:
            total += df.iloc[i]['판매수량']
    item_qty[item] = total

print("[ 품목별 총 판매수량 ]")
for item, qty in item_qty.items():
    print(item, ":", format(qty, ',') + "개")
print()

best_item  = items[list(item_qty.values()).index(max(item_qty.values()))]
worst_item = items[list(item_qty.values()).index(min(item_qty.values()))]
print("가장 많이 팔린 품목:", best_item, "(" + str(item_qty[best_item]) + "개)")
print("가장 적게 팔린 품목:", worst_item, "(" + str(item_qty[worst_item]) + "개)")
```

```python
# 카테고리별 평균 매출액
categories = list(df['카테고리'].unique())
category_sales = {}
category_count = {}

for category in categories:
    total = 0
    count = 0
    for i in range(len(df)):
        if df.iloc[i]['카테고리'] == category:
            total += df.iloc[i]['매출액']
            count += 1
    category_sales[category] = total
    category_count[category] = count

print("[ 카테고리별 평균 매출액 ]")
for category in categories:
    avg = category_sales[category] // category_count[category]
    print(category, "평균 매출액:", format(avg, ',') + "원")
```

```python
# 지점별 매출 순위
sorted_branch = dict(sorted(branch_sales.items(),
                            key=lambda x: x[1],
                            reverse=True))
print("\n[ 지점별 매출 순위 ]")
rank = 1
for branch, sales in sorted_branch.items():
    print("{}위: {}  {}원".format(rank, branch, format(sales, ',')))
    rank += 1
```

```python
# 카테고리별 매출 비중
grand_total = sum(category_sales.values())
print("\n[ 카테고리별 매출 비중 ]")
sorted_category = dict(sorted(category_sales.items(),
                              key=lambda x: x[1],
                              reverse=True))
for category, sales in sorted_category.items():
    ratio = round(sales / grand_total * 100, 1)
    print("{} : {}%".format(category, ratio))
```

```python
# 날짜별 총 매출 추이
dates = list(df['날짜'].unique())
date_sales = {}

for date in dates:
    total = 0
    for i in range(len(df)):
        if df.iloc[i]['날짜'] == date:
            total += df.iloc[i]['매출액']
    date_sales[date] = total

print("\n[ 날짜별 총 매출 추이 ]")
for date, sales in date_sales.items():
    print("{}: {}원".format(date, format(sales, ',')))
```
