```python
print("택시 요금 계산기")
distance = float(input("이동 거리를 입력하세요 (km): "))
time = int(input("현재 시각을 입력하세요 (0~23): "))

# 기본요금
base_fare = 4800

# 추가요금 계산 (2km 초과분을 100m 단위로)
if distance > 2:
    extra_meter = (distance - 2) * 1000        # 초과 거리를 m로 변환
    extra_unit = int(extra_meter / 100)        # 100m 단위로 변환 (버림)
    extra_fare = extra_unit * 100              # 100m당 100원
else:
    extra_fare = 0

# 심야할증
total_before = base_fare + extra_fare
if time >= 23 or time <= 4:
    night_fare = round(total_before * 0.2)
else:
    night_fare = 0

# 최종요금
total_fare = total_before + night_fare

print("=========================")
print("기본요금: ", base_fare, "원")
print("추가요금: ", extra_fare, "원")
print("심야할증: ", night_fare, "원")
print("최종요금: ", total_fare, "원")
print("=========================")
```

```python
print("■ RGB → HEX 변환기")
r = int(input("R 값 입력 (0~255): "))
g = int(input("G 값 입력 (0~255): "))
b = int(input("B 값 입력 (0~255): "))

hex_r = "%02X" % r
hex_g = "%02X" % g
hex_b = "%02X" % b

print("=========================")
print("RGB(%d, %d, %d)" % (r, g, b))
print("HEX 코드: #" + hex_r + hex_g + hex_b)
print("=========================")
```

```python
print("PIN 암호화 프로그램")
pin = input("4자리 PIN을 입력하세요: ")

d1 = int(pin[0])
d2 = int(pin[1])
d3 = int(pin[2])
d4 = int(pin[3])

e1 = (d1 + 7) % 10
e2 = (d2 + 7) % 10
e3 = (d3 + 7) % 10
e4 = (d4 + 7) % 10

c1 = chr(65 + e1)
c2 = chr(65 + e2)
c3 = chr(65 + e3)
c4 = chr(65 + e4)

print("=========================")
print("원본 PIN:   ", pin)
print("암호화 숫자:", str(e1) + str(e2) + str(e3) + str(e4))
print("암호화 문자:", c1 + c2 + c3 + c4)
print("=========================")
```

```python
print("픽셀 밝기 분석기")

print("--- 픽셀 1 ---")
r1 = int(input("R: "))
g1 = int(input("G: "))
b1 = int(input("B: "))

print("--- 픽셀 2 ---")
r2 = int(input("R: "))
g2 = int(input("G: "))
b2 = int(input("B: "))

print("--- 픽셀 3 ---")
r3 = int(input("R: "))
g3 = int(input("G: "))
b3 = int(input("B: "))

bright1 = round(r1 * 0.299 + g1 * 0.587 + b1 * 0.114, 2)
bright2 = round(r2 * 0.299 + g2 * 0.587 + b2 * 0.114, 2)
bright3 = round(r3 * 0.299 + g3 * 0.587 + b3 * 0.114, 2)

max_bright = max(bright1, bright2, bright3)
min_bright = min(bright1, bright2, bright3)

if max_bright == bright1:
    brightest = "픽셀 1"
elif max_bright == bright2:
    brightest = "픽셀 2"
else:
    brightest = "픽셀 3"

if min_bright == bright1:
    darkest = "픽셀 1"
elif min_bright == bright2:
    darkest = "픽셀 2"
else:
    darkest = "픽셀 3"

print("=========================")
print("픽셀 1 밝기:", bright1)
print("픽셀 2 밝기:", bright2)
print("픽셀 3 밝기:", bright3)
print()
print("가장 밝은 픽셀:", brightest)
print("가장 어두운 픽셀:", darkest)
print("=========================")
```
