```python
import pandas as pd
from scipy import stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd

df = pd.read_csv('week14_data.csv')
```

```python
#1
stats.ttest_1samp(df['midterm'], popmean=75)
# → t = -0.0171, p = 0.9864
```

```python
#2
stats.ttest_rel(df['midterm'], df['final'])
# → t = -8.6406, p < 0.001
# midterm 평균 = 74.985, final 평균 = 79.256
```

```python
#3
w = df[df['video_watched']=='Y']['final']
n = df[df['video_watched']=='N']['final']
stats.levene(w, n).pvalue        # → 0.2161 (등분산 OK)
stats.ttest_ind(w, n, equal_var=True)
# → t = 0.6348, p = 0.5271
```

```python
#4
groups = [df[df['study_method']==m]['final']
          for m in ['강의만','강의+실습','강의+실습+과제']]
stats.f_oneway(*groups)
# → F = 5.36, p = 0.006

pairwise_tukeyhsd(df['final'], df['study_method'], alpha=0.05)
# 강의+실습 vs 강의+실습+과제 : meandiff +4.83, p-adj=0.080, reject=False
# 강의+실습 vs 강의만          : meandiff -2.49, p-adj=0.441, reject=False
# 강의+실습+과제 vs 강의만     : meandiff -7.32, p-adj=0.004, reject=True
```

```python
#5
ct = pd.crosstab(df['major'], df['intern_experience'])
stats.chi2_contingency(ct)
# → χ² = 8.23, dof = 2, p = 0.0164
# 관측 빈도:
#               없음  있음
# 데이터사이언스   24   17
# 컴퓨터공학      14   14
# 통계           26    5     ← 인턴 적음
# 기대 빈도: 통계 인턴 있음 → 11.16명 (관측 5명)
```

```python
#6
stats.ttest_1samp(df['midterm'], popmean=70, alternative='greater')
# → t = 5.6912, p = 6.44e-08
```

```python
#7
stats.shapiro(df['midterm'])              # → p = 0.7969 (정규성 만족)
stats.shapiro(df['satisfaction_begin'])   # → p = 0.0019 (정규성 위배)
```

```python
#8
stats.ttest_rel(df['midterm'], df['final'], alternative='less')
# → t = -8.6406, p = 4.99e-14
```

```python
#9
ct = pd.crosstab(df['gender'], df['video_watched'])
stats.chi2_contingency(ct)
# → χ² ≈ 0.0, dof = (2-1)*(2-1) = 1, p ≈ 1.0
```

```python
#10
# (a) midterm
groups_m = [df[df['study_method']==m]['midterm']
            for m in ['강의만','강의+실습','강의+실습+과제']]
stats.f_oneway(*groups_m)  # → F = 11.84, p = 2.51e-05

# (b) final
groups_f = [df[df['study_method']==m]['final']
            for m in ['강의만','강의+실습','강의+실습+과제']]
stats.f_oneway(*groups_f)  # → F = 5.36, p = 0.006
```
