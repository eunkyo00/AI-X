```python
import random

print("<은행업무 순차형 프로세스 프로그램>")
print()

# 1단계: 번호표 발급
name    = input("고객 이름을 입력하세요: ")
ticket  = random.randint(300, 400)   # 번호표 자동 발급
current = random.randint(290, ticket)  # 현재 호출 번호 (번호표보다 작아야 함)
wait    = ticket - current

print()
print("[ 대기 안내 ]")
print("고객명     :", name)
print("내 번호표  :", ticket, "번")
print("현재 호출  :", current, "번")
print("대기 인원  :", wait, "명")
```

```python
print("<대학교 입학지원서 선택형 프로세스 프로그램>")
print()

# 1단계: 점수에 따라 대학교 선택
name  = input("지원자 이름을 입력하세요: ")
score = int(input("본인의 성적을 입력하세요 (1000점 만점): "))
print()

print("[ 점수 확인 ]")

if score >= 850:
    university = "A 대학교"
    cutline    = 850
elif score >= 800:
    university = "B 대학교"
    cutline    = 800
elif score >= 750:
    university = "C 대학교"
    cutline    = 750
else:
    university = "없음"
    cutline    = 0

# 2단계: 결과 출력
print("================================")
print("지원자      :", name)
print("본인 성적   :", score, "점")
print()

if university == "없음":
    print("⚠ 지원 가능한 대학교가 없습니다.")
    print("  (최소 커트라인: C 대학교 750점)")
else:
    print("지원 대학교 :", university)
    print("커트라인    :", cutline, "점")
    print("점수 여유   :", score - cutline, "점")
    print()
    print("[ 지원서 접수 ]")
    print("기간 내 입학지원서를 접수합니다.")
    print(university+"에 직접 방문하여 지원서를 접수합니다.")
    print()
    print(name+"님의 지원서 접수가 완료되었습니다.")

print("================================")
```

```python
print("<운전면허 학과 시험 반복형 프로세스 프로그램>")
print()

# 1단계: 면허 종류 선택
print("[ 운전면허 종류 안내 ]")
print("1종 보통: 승차정원 15명 이하 승합자동차 운전 가능")
print("2종 보통: 일반 승용차 및 승차정원 10인 이하 승합차 운전 가능")
print()

license_type = int(input("취득할 면허를 선택하세요 (1: 1종 보통 / 2: 2종 보통): "))

if license_type == 1:
    cutline   = 70
    type_name = "1종 보통"
else:
    cutline   = 60
    type_name = "2종 보통"

print()
print(type_name, "선택 완료 (커트라인:", cutline, "점)")
print()

# 2단계: 합격할 때까지 반복 응시
attempt = 0

while True:
    attempt += 1
    print("[ 제", attempt, "회 학과 시험 ]")
    score = int(input("시험 점수를 입력하세요: "))

    if score >= cutline:
        print("✅ 합격! (" + str(score) + "점 / 커트라인 " + str(cutline) + "점)")
        break
    else:
        print("❌ 불합격 (" + str(score) + "점 / 커트라인 " + str(cutline) + "점)")
        print("다음 시험을 준비합니다...\n")

# 3단계: 최종 결과
print()
print("================================")
print("면허 종류   :", type_name)
print("최종 점수   :", score, "점")
print("총 응시 횟수:", attempt, "회")
print()
print("학과 시험 합격! 코스 시험에 도전합니다. 🚗")
print("================================")
```

```python
import random

print("========================================")
print("       스마트 병원 접수 시스템")
print("========================================\n")

emergency_count = 0
general_count   = 0
total_fee       = 0
total_patients  = 0

while True:
    # 종료 조건
    name = input("환자 이름을 입력하세요 (-1: 종료): ")
    if name == "-1":
        break

    # 입력 받기
    age    = int(input("나이를 입력하세요: "))
    temp   = float(input("체온을 입력하세요 (°C): "))
    sys_bp = int(input("수축기 혈압을 입력하세요: "))
    dia_bp = int(input("이완기 혈압을 입력하세요: "))

    print("\n[ 증상 목록 ]")
    print("1. 두통/어지러움")
    print("2. 기침/호흡 곤란")
    print("3. 복통/소화불량")
    print("4. 근육통/관절통")
    print("5. 피부 발진/가려움")
    symptom = int(input("증상 번호를 선택하세요: "))

    # 응급 여부 판단
    is_emergency     = False
    emergency_reason = ""

    if temp > 39.0:
        is_emergency     = True
        emergency_reason = "체온 39.0도 초과"
    elif sys_bp > 180:
        is_emergency     = True
        emergency_reason = "수축기 혈압 180 초과"
    elif dia_bp > 110:
        is_emergency     = True
        emergency_reason = "이완기 혈압 110 초과"

    # 진료과 배정
    if is_emergency:
        department = "응급실"
    elif symptom == 1:
        department = "신경과"
    elif symptom == 2:
        department = "호흡기내과"
    elif symptom == 3:
        department = "소화기내과"
    elif symptom == 4:
        department = "정형외과"
    elif symptom == 5:
        department = "피부과"
    else:
        department = "일반내과"

    # 대기 번호 발급
    if is_emergency:
        emergency_count += 1
        ticket    = "EM-" + str(emergency_count).zfill(3)
        wait_time = 0
    else:
        general_count += 1
        ticket    = "P-" + str(general_count).zfill(3)
        wait_time = random.randint(5, 30)

    # 진료비 계산
    base_fee = 15000

    if is_emergency:
        fee = base_fee * 3
    else:
        fee = base_fee

    if age >= 65:
        discount      = round(fee * 0.3)
        discount_info = "65세 이상 30% 할인"
    elif age <= 15:
        discount      = round(fee * 0.2)
        discount_info = "15세 이하 20% 할인"
    else:
        discount      = 0
        discount_info = "없음"

    final_fee = fee - discount

    # 통계 누적
    total_patients += 1
    total_fee      += final_fee

    # 결과 출력
    print("\n----------------------------------------")
    print("[ 접수 결과 ]")
    print("대기 번호  :", ticket)
    print("환자 이름  :", name, "(" + str(age) + "세)")
    print("체온       : " + str(temp) + "°C")
    print("혈압       :", sys_bp, "/", dia_bp)

    if is_emergency:
        print("응급 여부  : 🚨 응급 (" + emergency_reason + ")")
    else:
        print("응급 여부  : ✅ 정상")

    print("배정 진료과:", department)

    if is_emergency:
        print("대기 시간  : 0분 (즉시 입실)")
    else:
        print("대기 시간  :", wait_time, "분")

    print("기본 진료비:", format(fee, ',') + "원")

    if discount > 0:
        print("할인 적용  : -" + format(discount, ',') + "원 (" + discount_info + ")")
    else:
        print("할인 적용  : 없음")

    print("최종 진료비:", format(final_fee, ',') + "원")
    print("----------------------------------------\n")

# 최종 통계 출력
print("========================================")
print("[ 오늘의 진료 통계 ]")
print("총 접수 환자  :", total_patients, "명")
print("응급 환자     :", emergency_count, "명")
print("일반 환자     :", general_count, "명")
print("총 진료비 합계:", format(total_fee, ',') + "원")

if total_patients > 0:
    avg_fee = round(total_fee / total_patients)
    print("평균 진료비   :", format(avg_fee, ',') + "원")

print("========================================")
```
