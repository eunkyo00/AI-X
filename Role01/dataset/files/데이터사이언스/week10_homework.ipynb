```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, OrdinalEncoder

df = pd.read_csv("week10_hospital_raw.csv")
```

```python
print(df.shape)
print(df.dtypes)
print(df.isnull().sum())
```

```python
#1
print(df.isnull().sum().sum())
```

```python
#2
missing = df.isnull().sum()
print(missing[missing > 0].sort_values(ascending=False))
```

```python
#3
df[df.isnull().any(axis=1)]
```

```python
#4
df["혈압"] = df["혈압"].fillna(df["혈압"].median())
print(round(df["혈압"].mean(), 1))
```

```python
#5
Q1 = df["나이"].quantile(0.25)
Q3 = df["나이"].quantile(0.75)
IQR = Q3 - Q1
print(IQR)
```

```python
#6
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df["나이"] < lower) | (df["나이"] > upper)]
print(len(outliers))
```

```python
#7
df_clean = df[(df["나이"] >= lower) & (df["나이"] <= upper)]
df_clean = df_clean.dropna(subset=["나이"])
print(round(df_clean["나이"].mean(), 1))
```

```python
#8
Q1_b = df["혈압"].quantile(0.25)
Q3_b = df["혈압"].quantile(0.75)
IQR_b = Q3_b - Q1_b

lower_b = Q1_b - 1.5 * IQR_b
upper_b = Q3_b + 1.5 * IQR_b
df["혈압"] = df["혈압"].clip(lower=lower_b, upper=upper_b)
```

```python
#9
df["진료비"] = df["진료비"].astype(str).str.replace(",","")
print(df["진료비"].dtype)
```

```python
df["진료비"] = df["진료비"].astype(str).str.replace(",", "").astype(int)
print(df["진료비"].dtype)
```

```python
#10
df["입원일"] = pd.to_datetime(df["입원일"], format="%Y-%m-%d")
#11
df["월"] = df["입원일"].dt.month
print((df["월"] == 1).sum())
```

```python
#12
oe = OrdinalEncoder(categories=[["BRONZE", "SILVER", "GOLD", "VIP"]])
df["등급_ord"] = oe.fit_transform(df[["등급"]])
print(df[df["등급"] == "GOLD"]["등급_ord"].iloc[0])
```

```python
df_ohe = pd.get_dummies(df, columns=["성별"], drop_first=True)
print([c for c in df_ohe.columns if "성별" in c])
```

```python
#15
df_sc = df[(df["나이"] >= lower) & (df["나이"] <= upper)].copy()
ss = StandardScaler()
df_sc[["나이_scaled"]] = ss.fit_transform(df_sc[["나이"]])
print(round(df_sc["나이_scaled"].mean(), 4))
```
